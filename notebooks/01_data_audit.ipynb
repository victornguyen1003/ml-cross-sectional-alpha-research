# Load data

```python
import yfinance as yf
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
headers = {"User-Agent": "Mozilla/5.0 (X11; Linux x86_64)"}

sp500_df = pd.read_html(url, storage_options=headers)[0]
sp500_tickers = sp500_df['Symbol'].str.replace('.', '-',regex=False).rename('ticker')

sp500_tickers
```

```python
sp500_path = Path('../config/sp500_tickers.csv')
sp500_path.parent.mkdir(parents=True, exist_ok=True)
sp500_tickers.to_csv(sp500_path, index=False)
```

```python
raw = yf.download(tickers=sp500_tickers.tolist(), start='2010-01-01', auto_adjust=False, actions=True, group_by='ticker')
raw
```

```python
raw_long = raw.copy().stack(level='Ticker').reset_index()
raw_long.columns.name = None
raw_long.columns = raw_long.columns.str.lower().str.replace(' ', '_', regex=False)
raw_long = raw_long.sort_values(by=['ticker', 'date'])
raw_long
```

```python
raw_path = Path('../data/raw/sp500_2010_2026.csv')
raw_path.parent.mkdir(parents=True, exist_ok=True)
raw_long.to_csv(raw_path, index=False)
```

```python
clean = raw_long.copy().drop_duplicates().dropna(subset='adj_close')
clean['month'] = clean['date'].dt.to_period('M')
clean = clean[clean['month'] < pd.Timestamp.today().to_period('M')]
clean = clean.groupby(by=['ticker', 'month']).tail(1).reset_index(drop=True)

prev_month = clean.groupby('ticker').shift(1)
month_gap = clean['month'].astype('int64') - prev_month['month'].astype('int64')

month_gap
```

```python
clean['return'] = clean['adj_close'] / prev_month['adj_close'] - 1
clean['return'] = np.where(month_gap!=1, np.nan, clean['return'])
clean['target_return'] = clean.groupby('ticker')['return'].shift(-1)

print("Check duplicates:")
print(clean.duplicated(['ticker','month']).sum())

print("\nCheck missing values:")
print(clean.isna().sum())

clean
```

```python
clean_path = Path('../data/interim/monthly_panel.csv')
clean_path.parent.mkdir(parents=True, exist_ok=True)
clean.to_csv(clean_path, index=False)
```
